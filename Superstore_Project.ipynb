```python
# Importing libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Read CSV file
df = pd.read_csv("SuperStore.csv", encoding="Unicode_escape")
```

```python
# Showing Data
df
```

# *Data understanding and Cleaning*

```python
df.describe()
```

```python
df.columns
```

```python
df.info()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"]= pd.to_datetime(df["Ship Date"])
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Month Name"] = df["Order Date"].dt.month_name()
df["Quarter"] = df["Order Date"].dt.quarter
df["Day of Week"] = df["Order Date"].dt.day_name()
```

```python
df.info()
```

# *Sales Performance*

```python
# Finding Total sales
Total_Sales = df["Sales"].sum()
print("Total Sales = ", Total_Sales)
```

```python
# Finding Total profit
Total_Profit = df["Profit"].sum()
print("Total Profit = ", Total_Profit)
```

```python
# Finding Total quantity sold
Quantity_Sold = df["Quantity"].sum()
print("Total Quantity Sold = ", Quantity_Sold)
```

```python
# Finding Average sales amount per order
Average_Sales = df.groupby("Order ID")["Sales"].sum().mean()
print("Average Sales per Order = ", Average_Sales)
```

```python
# Finding Year has the highest sales
Highest_Sales_Year = df.groupby("Year")["Sales"].sum().sort_values(ascending = False)
print("Highest Sales Year = ", Highest_Sales_Year)
```

```python
# Finding Month has the highest sales
Highest_Sales_Month = df.groupby("Month Name")["Sales"].sum().sort_values(ascending = False)
print("Highest Sales Month = ", Highest_Sales_Month)
```

```python
# Finding Sales increasing or decreasing overtime
Monthly_Sales_Trend = df.groupby("Month")["Sales"].sum()
Monthly_Sales_Trend.plot(
    kind = "line",
    figsize = (12,5),
    title = "Monthly Sales Trend",
    marker = "o",
    markerfacecolor= "white",
    markeredgecolor= "Red",
    ylabel = "Total Sales",
    xticks = (range(1,13)),
    grid =(True)
)
```

```python
# Finding which Quarter performs best
Highest_Sales_Quarter = df.groupby("Quarter")["Sales"].sum().sort_values(ascending = False)
print("Highest Sales Quarter = ", Highest_Sales_Quarter)
```

```python
# Finding the Day of Week has the highest sales
Highest_Sales_Day = df.groupby("Day of Week")["Sales"].sum().sort_values(ascending = False)
print("Day of Week has Highest Sales = ", Highest_Sales_Day)
```

# *Profit Analysis*

```python
# Finding Overall Profit margin 
Profit_Margin = round((Total_Profit/Total_Sales)*100,2)
print("Profit Margin = ", Profit_Margin)
```

```python
# Finding which Year generate the highest profit
Highest_Profit_Year = df.groupby("Year")["Profit"].sum().sort_values(ascending = False)
print("Highest Profit Year = ", Highest_Profit_Year)
```

```python
# Finding which Month generate the highest profit
Highest_Profit_Month = df.groupby("Month Name")["Profit"].sum().sort_values(ascending = False)
print("Highest Profit Month = ", Highest_Profit_Month)
```

```python
# Finding which categories are most profitable
categories = df.groupby("Category")["Profit"].sum().sort_values(ascending = False)
print("Most Profitable Categories = ", categories)
```

```python
# Finding which sub-categories are least profitable
Sub_categories = df.groupby("Sub-Category")["Profit"].sum().sort_values(ascending = True)
print("Least Profitable Sub-Categories = ", Sub_categories)
```

```python
# Finding which Products generate highest profit
Profitable_prod = df.groupby("Product Name")["Profit"].sum().sort_values(ascending = False)
print("Products Generated the Highest Profit = ", Profitable_prod)
```

```python
# Finding which Products are non-profitable and causes highest loss
Unprofitable_prod = df.groupby("Product Name")["Profit"].sum().sort_values(ascending = True).head(10)
print("Products Causes the Highest lose = ", Unprofitable_prod)
```

```python
# Finding that high sales always linked with high profit
Product_performance = df.groupby("Product Name")[["Sales", "Profit"]].sum().sort_values("Sales", ascending = False)
print("Product perfromance = ", Product_performance)
```

```python
# States are losing money
States = df.groupby("State")["Profit"].sum().sort_values().head(10)
print("State Losing Money = ", States)
```

```python
# Regions are losing money
Regions = df.groupby("Region")["Profit"].sum().sort_values()
print("Region Losing Money = ", Regions)
```

```python
# Finding relation between Discount and Profit
Discount_Profit = df.groupby("Discount")["Profit"].sum().sort_values(ascending = False)
print("Relation b/w Discount and Profit = ", Discount_Profit)
```

```python
# As the Discount increases then Profit decreases
Discount_Profit = df.groupby("Discount")["Profit"].sum()
Discount_Profit.plot(
    kind = "line",
    color = "Red",
    figsize = (12,6),
    title = "Relationship between Discount and Profit",
    marker = "o",
    markerfacecolor = "Yellow",
    markeredgecolor = "Black",
    grid = (True)
)
```

# *Shipping Analysis*

```python
# Which shipping mode is used the most?
Most_used = df["Ship Mode"].value_counts().sort_values(ascending = False)
Most_used
```

```python
# Which shipping mode generates the highest sales?
High_sales_shipping = df.groupby("Ship Mode")["Sales"].sum().sort_values(ascending = False)
print("Shipping mode generates highest sales : ", High_sales_shipping)
```

```python
# Which shipping mode generates the highest profit?
High_profit_shipping = df.groupby("Ship Mode")["Profit"].sum().sort_values(ascending = False)
print("Shipping mode generates highest profit : ", High_profit_shipping)
```

```python
# What is the average shipping time?
df["Shipping Day"] = (df["Ship Date"] - df["Order Date"]).dt.days
Avg_shiptime = round(df["Shipping Day"].mean())
print("Average Shipping time : ", Avg_shiptime, "Days")
```

```python
# Does shipping time affect profit?
affect_profit = df.groupby("Shipping Day")["Profit"].mean()
affect_profit.plot(
    title = "Relation between Shipping time and Profit",
    kind = "line",
    color = "Red",
    figsize = (12,6),
    marker = "o",
    markerfacecolor = "white",
    markeredgecolor = "Black",
    ylabel = "Profit",
    grid = (True)
)
plt.show() 
df[["Shipping Day", "Profit"]].corr()
```

```python
# Which shipping mode is fastest?
Fastest_mode = df.groupby("Ship Mode")["Shipping Day"].mean().sort_values()
print("Fastest Shipping Mode : ",round(Fastest_mode))
```

```python
# Which customer segment uses which shipping mode most?
Most_usedby_segments = df.groupby("Segment")["Ship Mode"].value_counts()
print("Customer segment uses most ship mode : ", Most_usedby_segments)
```

```python
# Are urgent shipping modes more profitable?
Urgent_ship = df.groupby("Ship Mode")["Profit"].sum().sort_values()
print("Urgent shipping is profitable ? : No",)
Urgent_ship
```

# *Discount Analysis*

```python
# What is the average discount?
Avg_disc = df["Discount"].mean()*100
print("Average Dsicount : ", round(Avg_disc),"%")
```

```python
# Which category receives the highest discount?
High_Disc_Category = df.groupby("Category")["Discount"].mean().sort_values(ascending = False)
print("Category recieves highest discount : ")
round(High_Disc_Category*100)
```

```python
# Which region receives the highest discount?
High_Disc_region = df.groupby("Region")["Discount"].mean().sort_values(ascending = False)
print("Region recieves highest discount : ")
round(High_Disc_region*100)
```

```python
# Does discount increase sales?
Discount_sales = df.groupby("Discount")["Sales"].mean()
Discount_sales.plot (
    kind = "line",
    title = "Discount and Sales",
    figsize = (12,6),
    marker = "o",
    markerfacecolor = "yellow",
    markeredgecolor = "red",
    ylabel = "Sales",
    grid = (True)
)
df[["Discount", "Sales"]].corr()
```

```python
# Does discount reduce profit?
Discount_profit = df.groupby("Discount")["Profit"].mean()
Discount_profit.plot(
    kind = "line",
    title = "Discount and Profit",
    figsize = (12,6),
    marker = "o",
    markerfacecolor = "yellow",
    markeredgecolor = "red",
    ylabel = "Profit",
    grid = (True)
)
df[["Discount", "Profit"]].corr()
```

```python
# What discount range causes loss?
Discount_loss = df.groupby("Discount")["Profit"].mean().sort_values()
print("Discount range causes loss : 30 to 80")
Discount_loss
```

```python
# Which products are heavily discounted?
Discount_products = df.groupby("Product Name")["Discount"].mean().sort_values(ascending = False).head(10)
print("Products are heavily discounted : ")
Discount_products*100
```
